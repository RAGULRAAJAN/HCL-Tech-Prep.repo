```python
import numpy as np

marks = np.array([95, 82, 74, 61, 45])
conditions = [
    marks >= 90,
    (marks >= 75) & (marks < 90),
    (marks >= 60) & (marks < 75),
    marks < 60
]
grades_list = ['A', 'B', 'C', 'F']
grades = np.select(conditions, grades_list)

print("Marks:", marks)
print("Classified Grades:", grades)
```

```python
import numpy as np

marks_matrix = np.array([
    [78, 85, 90], 
    [65, 72, 80], 
    [88, 91, 84], 
    [56, 62, 70], 
    [95, 89, 92]
])

# Individual metrics
total_marks = np.sum(marks_matrix, axis=1)
average_marks = np.mean(marks_matrix, axis=1)
highest_marks = np.max(marks_matrix, axis=1)
lowest_marks = np.min(marks_matrix, axis=1)

# Class metrics
class_average = np.mean(marks_matrix)
above_avg_mask = average_marks > class_average
students_above_avg = np.where(above_avg_mask)[0] + 1 

print("Marks Matrix:\n", marks_matrix)
print("Total Marks per Student:", total_marks)
print("Average Marks per Student:", average_marks)
print("Highest Mark per Student:", highest_marks)
print("Lowest Mark per Student:", lowest_marks)
print("Overall Class Average:", round(class_average, 2))
print("Students (1-5) performing above class average:", students_above_avg)
```
